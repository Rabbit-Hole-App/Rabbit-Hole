## Experiment 1 · Sigmoid
$$\sigma(x) = \frac{1}{1 + e^{-x}}$$
Change `score` below. Predict whether the result will be above or below 0.5, then run the cell.

```python
import math

score = 0
probability = 1 / (1 + math.exp(-score))
print(f'score = {score}; sigmoid = {probability:.4f}')
```

## Experiment 2 · From a reference point to a box
This simplified, decoded box example illustrates anchor-free geometry. It does not implement YOLO's learned head or its distribution decoding. All coordinates here use the same image coordinate system.

A reference point $(x,y)$ and distances $(l,t,r,b)$ give corners $(x-l,y-t)$ and $(x+r,y+b)$. Change the distances and inspect the rectangle.

```python
from IPython.display import SVG, display

x, y = 160, 120
left, top, right, bottom = 60, 40, 100, 70
assert min(left, top, right, bottom) >= 0, 'Distances must be non-negative'
x1, y1, x2, y2 = x-left, y-top, x+right, y+bottom
print('Box corners:', (x1, y1, x2, y2))
display(SVG(f'''<svg xmlns="http://www.w3.org/2000/svg" width="360" height="250" viewBox="0 0 360 250">
<rect width="360" height="250" fill="#fafafa"/>
<rect x="{x1}" y="{y1}" width="{x2-x1}" height="{y2-y1}" fill="#dbeafe" stroke="#2563eb" stroke-width="3"/>
<circle cx="{x}" cy="{y}" r="5" fill="#ef4444"/>
</svg>'''))
```
